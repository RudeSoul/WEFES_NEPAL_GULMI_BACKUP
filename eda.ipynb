```python
import matplotlib.pyplot as plt
```

```python
pho = df['phosphorus'].values
nit = df['nitrogen'].values
pot = df['potassium'].values
ph = df['ph'].values
fig, axs = plt.subplots(2, 2, figsize=(10, 10))
axs[0, 0].hist(nit, bins=100)
axs[0,0].set_title('Nitrogen Distribution')
axs[0, 1].hist(pot, bins=100)
axs[0, 1].set_title('Potassium Distribution')
axs[1, 0].hist(ph, bins=100)
axs[1, 0].set_title('pH Distribution')
axs[1, 1].hist(pho, bins=100)
axs[1, 1].set_title('Phosphorus Distribution')
fig.show()
```

```python
fig, axs = plt.subplots(2, 2, figsize=(10, 10))
axs[0, 0].boxplot(nit)
axs[0,0].set_title('Nitrogen Distribution')
axs[0, 1].boxplot(pot)
axs[0, 1].set_title('Potassium Distribution')
axs[1, 0].boxplot(ph)
axs[1, 0].set_title('pH Distribution')
axs[1, 1].boxplot(pho)
axs[1, 1].set_title('Phosphorus Distribution')
fig.show()
```

```python
# df.groupby('soil_type')['phosphorus'].mean().sort_values(ascending=False).plot(kind='bar')
fig, axs = plt.subplots(2, 2, figsize=(10, 10))
axs[0, 0].bar(df.groupby('soil_type')['nitrogen'].mean().sort_values(ascending=False).index, df.groupby('soil_type')['nitrogen'].mean().sort_values(ascending=False).values)
axs[0,0].set_title('Nitrogen Distribution')
axs[0, 1].bar(df.groupby('soil_type')['potassium'].mean().sort_values(ascending=False).index, df.groupby('soil_type')['potassium'].mean().sort_values(ascending=False).values)
axs[0, 1].set_title('Potassium Distribution')
axs[1, 0].bar(df.groupby('soil_type')['ph'].mean().sort_values(ascending=False).index, df.groupby('soil_type')['ph'].mean().sort_values(ascending=False).values)
axs[1, 0].set_title('pH Distribution')
axs[1, 1].bar(df.groupby('soil_type')['phosphorus'].mean().sort_values(ascending=False).index, df.groupby('soil_type')['phosphorus'].mean().sort_values(ascending=False).values)
axs[1, 1].set_title('Phosphorus Distribution')
for ax in axs.flatten():
    # Rotate text 45 degrees and align the edge to the tick mark
    ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right')
plt.tight_layout()
fig.show()

```
