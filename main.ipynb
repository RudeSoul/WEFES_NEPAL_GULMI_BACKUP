```python
import geopandas as gpd
import pandas as pd
from shapely.geometry import Point

# 1. Load the Nepal districts JSON (GeoJSON) file
# GeoPandas automatically parses features, geometry, and coordinates
districts_gdf = gpd.read_file("data/geojson/nepal-districts.json")

# 2. Load your CSV file using pandas
csv_df = pd.read_csv("data/geojson/soildataNepal.csv")

# 3. Convert CSV lat/lon coordinates into Shapely Point geometries
# Note: Coordinate order must be (Longitude, Latitude) to match GeoJSON standard
geometry = [Point(xy) for xy in zip(csv_df["lon"], csv_df["lat"])]

# 4. Create a GeoDataFrame from the CSV data
# WGS84 (EPSG:4326) is the standard coordinate system for GPS lat/lon
points_gdf = gpd.GeoDataFrame(csv_df, geometry=geometry, crs="EPSG:4326")


# 5. Ensure the district boundaries use the exact same coordinate system
if districts_gdf.crs != "EPSG:4326":
    districts_gdf = districts_gdf.to_crs("EPSG:4326")

# 6. Perform a spatial join (sjoin) to find which point is 'within' which district
# 'how="left"' keeps all rows from your original CSV even if they fall outside Nepal
result_gdf = gpd.sjoin(points_gdf, districts_gdf, how="left", predicate="within")

# 7. Drop the temporary geometry column and save back to a clean CSV
final_df = pd.DataFrame(result_gdf).drop(columns=["geometry", "index_right"])
final_df.to_csv("mapped_output.csv", index=False)

print(final_df.head())
```

```python
import pandas as pd
import requests
from io import StringIO

def download_nepal_hydro_data():
    print("Connecting to the database...")
    
    # The URL containing the comprehensive table of Nepal's power stations
    url = "https://en.wikipedia.org/wiki/List_of_power_stations_in_Nepal"
    
    # We use requests to get the HTML content securely
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"
    }
    response = requests.get(url, headers=headers)
    
    if response.status_code == 200:
        print("Data found! Extracting tables...")
        
        # FIX: Wrap the HTML text in StringIO so Pandas processes it correctly
        html_data = StringIO(response.text)
        
        # Read the tables from the StringIO object
        tables = pd.read_html(html_data)
        
        # The first table on this specific page is the "Hydroelectric stations" table
        hydro_df = tables[0]
        
        # Clean up the table (Drop empty/reference columns if any)
        if 'Refs' in hydro_df.columns:
            hydro_df = hydro_df.drop(columns=['Refs'])
            
        # Clean up column names for better readability in CSV
        hydro_df.columns = [col.replace(' ', '_').replace('(', '').replace(')', '') for col in hydro_df.columns]
        
        # Preview the first few rows
        print("\n--- Data Preview ---")
        print(hydro_df[['Name_of_station', 'Capacity_MW', 'Owner']].head())
        print("--------------------\n")
        
        # Save to CSV
        filename = "nea_hydro_complete_database.csv"
        hydro_df.to_csv(filename, index=False)
        print(f"Success! The full dataset with {len(hydro_df)} hydropower stations has been saved as '{filename}'.")
        
    else:
        print(f"Failed to fetch data. HTTP Status Code: {response.status_code}")

# Run the function
if __name__ == "__main__":
    download_nepal_hydro_data()
```
